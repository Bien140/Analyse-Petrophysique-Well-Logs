```python
# !/usr/bin/env python3
from db_connection import get_engine
import pandas as pd

# 1. On se connecte à PostgreSQL
engine = get_engine()

try:
    with engine.connect() as conn:
        print("Connexion réussie à PostgreSQL")
        
        # 2. On lit le fichier CSV nettoyé
        df_clean = pd.read_csv("well_logs_clean.csv", encoding='utf-8')
        
        # 3. On envoie les données dans PostgreSQL
        df_clean.to_sql('well_logs', engine, if_exists='replace', index=False)
        print("SUCCÈS !",len(df_clean),"lignes importées dans la table 'well_logs'.")
        
except Exception as e:
    print("Échec de la connexion ou de l'importation :", e)
```
